```python
from abc import ABC, abstractmethod

##Abstract Base Class
class Vehicle(ABC):
    def drive(self):
        print("the vehicle is used to travel")
    
    @abstractmethod
    def start_engine(self):
        pass

class Car(Vehicle):
    def start_engine(self):
        print("the car engine started")

def operate_vehicle(vivek):
    vivek.start_engine()

car=Car()
operate_vehicle(car)
```

Final Takeaway

Vehicle = blueprint (abstract)

start_engine() = must be implemented

Car = real implementation

operate_vehicle() = uses abstraction + polymorphism
